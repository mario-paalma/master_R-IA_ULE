## Sección 2 — Operaciones con tensores

### Ejercicio 2.1 — Aritmética y reducciones

Dados `x` e `y`, calcula:

- `suma`      : suma elemento a elemento  
- `producto`  : producto elemento a elemento (Hadamard)  
- `potencia`  : `x` al cuadrado  
- `norm_x`    : `x` normalizado (media 0, std 1) usando sólo operaciones tensoriales  
- `max_val`   : valor máximo de `x` (escalar Python, usa `.item()`)  
- `argmax_x`  : índice del valor máximo

````python
x = torch.tensor([1., 2., 3., 4., 5.])
y = torch.tensor([5., 4., 3., 2., 1.])

# --- Tu código aquí ---
suma     = x+y
producto = x*y
potencia = x**2
norm_x   = (x-x.mean())/(x.std())
max_val  = x.max().item()
argmax_x = x.argmax().item()
````

````python
assert suma.tolist() == [6., 6., 6., 6., 6.]
assert producto.tolist() == [5., 8., 9., 8., 5.]
assert potencia.tolist() == [1., 4., 9., 16., 25.]
assert abs(norm_x.mean().item()) < 1e-6 and abs(norm_x.std().item() - 1.0) < 1e-5
assert max_val == 5.0
assert argmax_x == 4
print("✓ Ejercicio 2.1 OK")
````

### Ejercicio 2.2 — Álgebra matricial

Dadas `A` (3×4) y `B` (4×2):

- `C`        : producto matricial A @ B (usa `torch.matmul`)  
- `A_T`      : transpuesta de A  
- `traza`    : traza de A @ Aᵀ (resultado escalar Python)  
- `A_norm_r` : norma L2 de cada fila de A (forma `(3,)`)

````python
A = torch.arange(12, dtype=torch.float32).reshape(3, 4)
B = torch.ones(4, 2)

# --- Tu código 
C        = torch.matmul(A,B)
A_T      = A.T
traza    = torch.trace( torch.matmul(A,A_T)).item()
A_norm_r = torch.linalg.norm(A,dim=1)
````

````python
assert C.shape == (3, 2)
assert A_T.shape == (4, 3)
assert abs(traza - (A @ A.T).trace().item()) < 1e-3
assert A_norm_r.shape == (3,)
print("✓ Ejercicio 2.2 OK")
````

### Ejercicio 4.2 — Normalización batch con broadcasting

Dado un mini-batch `X` de forma `(batch, features)`, normaliza **cada feature** (columna) para que tenga media 0 y std 1 usando broadcasting (sin bucles).

- `mu`    : media por feature, forma `(1, features)`  
- `sigma` : desv. típica por feature, forma `(1, features)`  
- `X_norm`: X normalizado

````python
torch.manual_seed(0)
X = torch.randn(32, 16) * 3 + 2   # batch=32, features=16

# --- Tu código aquí ---
mu     = X.mean(dim=0,keepdim=True)
sigma  = X.std(dim=0,keepdim=True,unbiased=False)
X_norm = (X-mu)/sigma
````

## Sección 5 — `requires_grad` y gradientes básicos

### Ejercicio 5.1 — Derivada de una función escalar

Calcula el gradiente de $f(x) = 3x^2 + 2x - 1$ en $x = 2.0$ usando Autograd.

El gradiente analítico es $f'(x) = 6x + 2$, que en $x = 2$ vale **14**.

1. Crea `x = 2.0` con `requires_grad=True`.  
2. Evalúa `y = f(x)`.  
3. Llama a `y.backward()`.  
4. Extrae el gradiente en `grad_x`.

````python
# --- Tu código aquí ---
x      = torch.tensor(2.0,requires_grad=True)
y      = 3*(x**2) + 2*x - 1
# backward ...
y.backward()
grad_x = x.grad   # x.grad
````

````python
assert abs(y.item() - 15.0) < 1e-5, f"f(2) = {y.item()}, esperado 11"
assert abs(grad_x.item() - 14.0) < 1e-5, f"f'(2) = {grad_x.item()}, esperado 14"
print(f"f(2) = {y.item():.2f}   f'(2) = {grad_x.item():.2f}")
print("✓ Ejercicio 5.1 OK")
````

### Ejercicio 5.2 — Gradiente de la pérdida MSE

Dado un modelo lineal $\hat{y} = Xw$, calcula la pérdida MSE y su gradiente respecto a `w`.

$$L = \frac{1}{N}\sum_i(\hat{y}_i - y_i)^2$$

````python
torch.manual_seed(7)
X      = torch.randn(10, 3)
y_true = torch.randn(10)
w      = torch.randn(3, requires_grad=True)

# --- Tu código aquí ---
y_pred = torch.matmul(X,w)   # X @ w
loss   = torch.sum((y_pred-y_true)**2) * 1/(X.size(0))   # MSE escalar
# backward ...
loss.backward()
grad_w = w.grad   # w.grad
````

````python
# Verificación con diferencias finitas
eps = 1e-5
grad_fd = torch.zeros(3)
with torch.no_grad():
    for i in range(3):
        w_p = w.detach().clone(); w_p[i] += eps
        w_m = w.detach().clone(); w_m[i] -= eps
        grad_fd[i] = (((X@w_p - y_true)**2).mean() - ((X@w_m - y_true)**2).mean()) / (2*eps)
assert torch.allclose(grad_w, grad_fd, atol=1e-1), f"Gradiente incorrecto\nAutograd: {grad_w}\nFD: {grad_fd}"
print(f"Loss = {loss.item():.4f}")
print("✓ Ejercicio 5.2 OK")
````

### Ejercicio 5.3 — Paso manual de descenso de gradiente

Con el gradiente calculado en 5.2, realiza **un paso de SGD** con `lr = 0.1`:
$$w \leftarrow w - \text{lr} \cdot \nabla_w L$$

Comprueba que la loss disminuye. Usa `torch.no_grad()` para no crear grafo.  
Después de actualizar `w`, **pon a cero** el gradiente (`w.grad.zero_()`).

Finalmente, recalcula la loss con los nuevos pesos y guárdala en `loss_nueva`.

````python
lr = 0.1

# --- Tu código aquí ---
# 1. Actualiza w dentro de torch.no_grad()
# 2. Pon a cero el gradiente
# 3. Recalcula la loss
with torch.no_grad():
    w -=lr * grad_w

w.grad.zero_()

with torch.no_grad():
    y_pred = torch.matmul(X,w)
    loss_nueva = 1/(X.size(0)) * torch.sum((y_pred-y_true)**2)
````

````python
with torch.no_grad():
    loss_check = ((X @ w - y_true)**2).mean()
assert torch.allclose(loss_nueva, loss_check, atol=1e-5)
assert loss_nueva.item() < loss.item(), f"La loss no bajó: {loss.item():.4f} → {loss_nueva.item():.4f}"
print(f"Loss: {loss.item():.4f} → {loss_nueva.item():.4f}")
print("✓ Ejercicio 5.3 OK")
print("\n🎉 ¡Práctica P1 completada!")
````
